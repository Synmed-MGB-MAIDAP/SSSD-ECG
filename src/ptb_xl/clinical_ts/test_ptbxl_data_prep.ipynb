In this notebook, we look at the order of labels.

```python
import pickle
import numpy as np
from pathlib import Path
target_root =  Path("/home/nutansahoo/DiffuSETS_Exp/prerequisites") 
```

```python
# udf
def multihot_encode(x, num_classes):
    res = np.zeros(num_classes, dtype=np.float32)
    for y in x:
        res[y]=1
    return res
```

```python
df = pickle.load(open(target_root/("df_memmap"+".pkl"), "rb"))
lbl_itos = pickle.load(open(target_root/("lbl_itos"+".pkl"), "rb"))
```

```python
# label order
print (lbl_itos['label_all'])
```

```python
# create binary 71 len labels
ptb_xl_label_demographcis = ["label_all"]
df["label"] = df.apply(
    lambda row: np.concatenate([
        multihot_encode(row[label+"_numeric"], len(lbl_itos[label]))
        for label in ptb_xl_label_demographcis
    ]),
    axis=1
)
```

```python
# confirm the order of labels 
# return 1 if the sublist contains N=3
df.loc[df['label_all_numeric'].apply(lambda x: True if 3 in x else False), ['label_all', 'label_all_numeric', 'label']]
# This confirms that ABQRS is at index 3 in the order of labels
```

```python
# df.loc[df['report']==' ']
# use regex to convert all spaces to empty string in report column
df['report'] = df['report'].str.replace(r'\s+', '', regex=True)
```
